#### 配置选择路径

```python
import os
import ipywidgets as widgets
from IPython.display import display
# 下拉框
path = widgets.Dropdown(
    options=['work1', 'work2', 'work2_1'],
    description='选择路径：', disabled=False
    )
display(path)
# 选中后触发
os.chdir('./'+path.value)
work = os.getcwd()
print(work)
```
